# TimeSeries - Reindexing and Resampling
Using our stocks data again

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_pickle('stocks.pkl')
df.head()
```

In our current data, we have business days only. What if we want the ability to look up any day in our date range and get the value? We can look back to the last trading day in this case.

To keep this simple, let's look at AAL again.

```python
aal = df.loc['AAL']
aal
```

```python
start, end = aal.index.min(), aal.index.max()
print(start, end)
```

```python
new_index = pd.date_range(start, end)
new_index
```

```python
aal2 = aal.reindex(new_index, method='ffill')
aal2
```

Great, now every date has a time. But what if we wanted to do this to all secrurities? and let's only fill within two days to cater to a weekend, but not to something more dreastic!
